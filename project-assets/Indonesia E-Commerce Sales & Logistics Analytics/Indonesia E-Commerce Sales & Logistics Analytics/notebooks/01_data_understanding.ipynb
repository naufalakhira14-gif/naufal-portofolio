## Check Data Type

### Data Type Assessment

The dataset contains **9 integer variables and 10 object variables**.

Most numerical variables have already been interpreted correctly as integers. However, 
`Waktu Pesanan Dibuat` is currently stored as an `object`, even though it represents a timestamp. 
This variable should therefore be converted into a proper datetime format during data cleaning.

Two variables have fewer non-null observations than the total number of rows:

- `Alasan Pembatalan`
- `Waktu Pesanan Dibuat`

These missing values require further investigation before deciding whether they should be retained, transformed, or removed.

## Check Missing Value

```python
missing = pd.DataFrame({
    "missing_count": df.isnull().sum(),
    "missing_percentage": (df.isnull().mean() * 100).round(2)
})

missing[missing["missing_count"] > 0].sort_values(
    "missing_count",
    ascending=False
)
```

### Missing Value Assessment

Two columns contain missing values.

`Alasan Pembatalan` contains **18,018 missing values (86.43%)**. However, this does not necessarily indicate poor data quality. Since cancellation reasons are only relevant for cancelled orders, missing values may represent orders that were successfully completed or had another non-cancelled status.

`Waktu Pesanan Dibuat` contains **1,980 missing values (9.50%)**. Unlike cancellation reasons, order timestamps are expected to exist for every transaction. Therefore, these missing timestamps represent a more significant data quality issue and should be investigated further during the cleaning stage.

No missing-value treatment is performed at this stage because the purpose of this notebook is to understand the raw dataset before modifying it.

### Numerical Distribution Assessment

The descriptive statistics indicate that several numerical variables are strongly right-skewed.

For example:

- The median order quantity is **1 item**, while the maximum reaches **256 items**.
- The median order weight is **500 grams**, while the maximum reaches **375,000 grams**.

Further investigation using distribution analysis and outlier detection will be conducted during the cleaning and exploratory data analysis stages.

## Cardinality

```python
cardinality = pd.DataFrame({
    "unique_values": df.nunique(),
    "data_type": df.dtypes
}).sort_values("unique_values")

cardinality
```

### Cardinality Assessment

Cardinality analysis provides additional information about the role of each variable.

`order_id` has the highest cardinality with **20,848 unique values**, confirming its role as the unique transaction identifier.

`Waktu Pesanan Dibuat` also has high cardinality with **18,591 unique non-null timestamps**, which is expected for transactional data.

Geographic variables have moderate cardinality, with **34 provinces and 424 cities/regencies**, making them suitable for regional performance analysis.

Several categorical variables such as order status, payment method, cancellation reason, and shipping option have relatively low cardinality and can be effectively used for grouping, comparison, and dashboard filtering.

Meanwhile, `product_categories` has relatively high cardinality because individual orders may contain different combinations of product categories. This variable may require additional transformation if category-level analysis is needed.

# Data Understanding Conclusion

The initial assessment shows that the dataset contains **20,848 unique e-commerce orders and 19 variables**, covering sales, products, payments, logistics, geographic locations, returns, cancellations, and transaction timestamps.

Overall, the dataset provides sufficient dimensions for developing an e-commerce sales and logistics analytics project.

1. Column names use inconsistent naming conventions and should be standardized.
2. `Waktu Pesanan Dibuat` is stored as an object and should be converted to datetime.
3. The order timestamp contains **1,980 missing values (9.50%)** that require further investigation.
4. Missing values in `Alasan Pembatalan` appear to be structurally related to non-cancelled orders and should not be removed automatically.
5. No duplicated rows or duplicated order IDs were identified.
6. Several numerical variables contain extreme values that require further investigation before being classified as outliers.
7. Product categories may contain multiple categories within a single transaction and may require transformation for category-level analysis.
8. Categorical variables should be inspected for inconsistent labels before analysis.

Based on these findings, the next stage of the project will focus on **data cleaning and transformation** while preserving the integrity of the original raw dataset.
